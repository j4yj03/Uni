```python
#### params = 2000

range1 = range(1, params, 1)
range2 = range(1, params, 1)
range3 = range(1, params, 1)
range4 = range(1, params, 1)

y1_true = False
y2_true = False

def foo(x, a, b, c, d):
    return -(a/b)*x - (c/d)


print('start')
def run():
    for r1 in range1:
        for r2 in range2:
            
            if r2 <= r1:
                break
                
            for r3 in range3:
                for r4 in range4:
                    
                    if r4 <= r3:
                        break
    
                    y1_true = False
                    y2_true = False
    
                    y1 = foo(8, r1, r2, r3, r4)
                    y2 = foo(-5, r1, r2, r3, r4)
    
                    if y1 == -47:
                        y1_true = True
                        #print('found y1')
    
    
                    if y2 == 44:
                        y2_true = True
                        #print('found y2')
    
                    if y1_true and y2_true:
                        print('found y1 and y2')
                        print(f'-({r1}/{r2})*x - ({r3}/{r4})')
                        return


run()
```

```python
def a1(x):
    return 2*x

def a2(x):
    return x**4 + x**2

def a3(x):
    return np.abs(x)

def a4(x):
    return x**3


fig = plt.figure(figsize=(20, 10))
gs = gridspec.GridSpec(2, 2)


fn_list = [a1, a2, a3, a4]
ax_list = [plt.subplot(gs[n]) for n in range(4)]

try:
    for ax, fn, col in zip(ax_list, fn_list, list(mcolors.TABLEAU_COLORS)):
        for n in np.arange(-25, 25, 0.2):
            y = fn(n)
            ax.set_title(f'function: {fn.__name__}')
            ax.scatter(n, y, s=3, color=col)

    #plt.legend()
    plt.show()

except Exception as e:
    print(e)
```

```python
x = np.array([-5, -3, -1, 4])

y = np.array([74, 20, -1, 45])


fig = plt.figure(figsize=(10, 10))
plt.plot(x, y)
plt.show()

x_extended = np.c_[np.ones(4), x**2]

#print(x_extended)

M = x_extended.T.dot(x_extended)
r = x_extended.T.dot(y)

thetas = np.linalg.solve(M, r)
#np.linalg.inv(X_ext.T.dot(X_ext)).dot(X_ext.T).dot(y)

print(M)
print()
print(r)
print()
print(thetas)

x_test = np.arange(-6, 6, 0.05)

x_test_extended = np.c_[np.ones(x_test.shape[0]), x_test**2]

y_predict = x_test_extended.dot(thetas)

fig = plt.figure(figsize=(10, 10))
plt.plot(x,y)

plt.scatter(x_test, y_predict,  s=3, color='red')
   
plt.show()
```
